## The problem


You are given an integer array `nums` sorted in non-decreasing order (each value is greater than or equal to the one before it). Remove the duplicates in place so each distinct value appears once, keeping their order. Return `k`, the number of distinct values.

- The first `k` positions of `nums` must hold the distinct values, in their original order.
- Whatever is left after position `k` doesn’t matter.
- In place means no second array: only O(1) extra memory.

The judge checks both the returned `k` and the first `k` elements of `nums`.

| Input | Output | First k elements after the call |
|---|---|---|
| `nums = [1, 1, 2]` | `2` | `[1, 2]` |
| `nums = [0, 0, 1, 1, 1, 2, 2, 3, 3, 4]` | `5` | `[0, 1, 2, 3, 4]` |

Input limits: 1 ≤ length of `nums` ≤ 30,000. Values are between −100 and 100. The array is sorted.

Ask before you code: Is it definitely sorted? Is it fine to leave leftover values after position `k`?

## Solution: two pointers, same direction


Two pointers, same direction (the best answer, and the one to present in the interview):

Because the array is sorted, copies of a value always sit next to each other. Use two pointers in the same-direction shape:

- `read` visits every element after the first, left to right.
- `write` marks where the next new value should go.

Compare each `read` value with the last value you kept, `nums[write - 1]`. If it’s different, it’s a new value: copy it to `nums[write]` and move `write` forward. If it’s the same, skip it. Everything before `write` is the finished answer.

The first element is always kept, so `write` starts at 1.

```python
def remove_duplicates(nums):                  # O(n) time, O(1) memory
    if not nums:
        return 0
    write = 1                                 # nums[:write] is always the distinct prefix
    for read in range(1, len(nums)):
        if nums[read] != nums[write - 1]:     # a value we haven't kept yet
            nums[write] = nums[read]
            write += 1
    return write
```

```python
nums = [1, 1, 2, 2, 3]
```

```python
k = remove_duplicates(nums)
```

```python
k
```

```python
nums[:k]
```

## Careful: only the first k count


```python
nums = [1, 1, 2, 2, 3]
```

```python
k = remove_duplicates(nums)
```

```python
nums
```

```python
nums[:k]
```

Return 3. The answer is the first three elements, `[1, 2, 3]`. The trailing `[2, 3]` is leftover and ignored.

## Both compared


| Approach | Time | Extra memory | Note |
|---|---|---|---|
| `sorted(set(nums))` | O(n log n) | O(n) | Breaks the in-place rule |
| Two pointers, same direction | O(n) | O(1) | Default answer |
