## The problem


You are given two integer arrays `nums1` and `nums2`. Return the values that appear in both. Each value in the answer must be unique, and the order doesn’t matter.

| Input | Output | Why |
|---|---|---|
| `nums1 = [1, 2, 2, 1]`, `nums2 = [2, 2]` | `[2]` | 2 is in both; it appears once in the answer |
| `nums1 = [4, 9, 5]`, `nums2 = [9, 4, 9, 8, 4]` | `[9, 4]` | Any order is accepted |

Input limits: 1 ≤ length of each ≤ 1,000. Values between 0 and 1,000.

Ask before you code: Does the order of the answer matter? Should repeats be kept? Are the arrays sorted?

## Solution 1: brute force


Brute force: for each value in `nums1`, scan `nums2` for it and scan the answer so far to avoid repeats. Both scans are O(length), so this is O(n · m).

```python
def intersection_brute(nums1, nums2):         # O(n * m) time
    result = []
    for x in nums1:
        if x in nums2 and x not in result:    # each `in` scans a list
            result.append(x)
    return result
```

```python
intersection_brute([4, 9, 5], [9, 4, 9, 8, 4])
```

```python
intersection_brute([1, 2, 2, 1], [2, 2])
```

## Solution 2: sets


Sets (the best answer, and the one to present in the interview): first the idea spelled out, then the one-liner you’d actually write.

Turn `nums1` into a set. That gives you its distinct values and an instant “is this in `nums1`?” check. Then walk `nums2`: every value that’s in the set goes into the answer. Collect the answer in a set too, so a value that appears twice in `nums2` is only added once.

```python
def intersection_loop(nums1, nums2):          # O(n + m) time, O(n) memory
    in_first = set(nums1)
    result = set()
    for x in nums2:
        if x in in_first:
            result.add(x)
    return list(result)


def intersection(nums1, nums2):               # same thing in one line
    return list(set(nums1) & set(nums2))
```

```python
intersection_loop([4, 9, 5], [9, 4, 9, 8, 4])
```

```python
intersection([4, 9, 5], [9, 4, 9, 8, 4])
```

```python
intersection_loop([1, 2, 2, 1], [2, 2])
```

```python
intersection([1, 2, 2, 1], [2, 2])
```

## Solution 3: sort, then two pointers


Sort both, then two pointers: walk both sorted arrays together, always moving the pointer at the smaller value. Equal values are a match; skip repeats as you go. Because both arrays are sorted, the smaller value can’t match anything from the other pointer onward, which holds only values at least as big, so it is safe to skip. Matches come out in sorted order, so a repeat is always equal to the last value added, and comparing with `result[-1]` is enough to skip it.

```python
def intersection_sorted(nums1, nums2):        # O(n log n + m log m) time
    a, b = sorted(nums1), sorted(nums2)
    i = j = 0
    result = []
    while i < len(a) and j < len(b):
        if a[i] < b[j]:
            i += 1
        elif a[i] > b[j]:
            j += 1
        else:
            if not result or result[-1] != a[i]:   # skip repeats
                result.append(a[i])
            i += 1
            j += 1
    return result
```

```python
intersection_sorted([4, 9, 5], [9, 4, 9, 8, 4])
```

```python
intersection_sorted([1, 2, 2, 1], [2, 2])
```

## All three compared


| Approach | Time | Extra memory | When to use it |
|---|---|---|---|
| Brute force | O(n · m) | O(1) beyond the answer | Only to state as the starting point |
| Sort + two pointers | O(n log n + m log m) | O(n + m) as written, since `sorted()` makes new lists; O(1) beyond the answer if you may sort in place | Arrays already sorted, or memory is tight and you may sort in place |
| Sets | O(n + m) | O(n) for the loop version, which keeps a set of `nums1` and an answer of at most that size; O(n + m) for the one-liner, which builds a set of each array | Default answer |

## Follow-ups


| They ask | Your answer |
|---|---|
| “Both arrays are already sorted.” | Sort + two pointers, skipping the sort: O(n + m) time and no sets. |
| “Keep repeats: a value appearing twice in both should appear twice in the answer.” (LeetCode 350) | Now you need counts, not just presence: a frequency map. Code below. |
| “One array is tiny and the other is huge.” | Build the set from the small one, then stream the big one past it. Memory is O(size of the small array). |
| “`nums2` is on disk and too big for memory.” | Same: keep the set of `nums1` in memory and read `nums2` in chunks. |
| “Return the answer in the order values appear in `nums2`.” | Walk `nums2`, and add a value only the first time you see it (a seen-set). |

```python
from collections import Counter

def intersect_with_counts(nums1, nums2):      # LeetCode 350: keep repeats
    count = Counter(nums1)
    result = []
    for x in nums2:
        if count[x] > 0:                      # nums1 still has a copy to match
            result.append(x)
            count[x] -= 1
    return result
# intersect_with_counts([1, 2, 2, 1], [2, 2]) -> [2, 2]
```

```python
intersect_with_counts([1, 2, 2, 1], [2, 2])
```
