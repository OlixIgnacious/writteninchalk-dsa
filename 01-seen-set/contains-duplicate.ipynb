## The problem


You are given an integer array `nums`. Return `true` if any value appears at least twice in the array. Return `false` if every element is different.

| Input | Output | Why |
|---|---|---|
| `nums = [1, 2, 3, 1]` | `true` | 1 appears at index 0 and again at index 3 |
| `nums = [1, 2, 3, 4]` | `false` | All four values are different |
| `nums = [1, 1, 1, 3, 3, 4, 3, 2, 4, 2]` | `true` | Several values repeat |

Input limits: 1 ≤ length of `nums` ≤ 100,000. Each value is between −10⁹ and 10⁹.

Ask before you code: May I change the input array? Is memory limited?

## Solution 1: brute force


Brute force: compare every pair. Correct, but 100,000 items means about 5 billion comparisons.

```python
def contains_duplicate_brute(nums):           # O(n^2) time, O(1) space
    n = len(nums)
    for i in range(n):
        for j in range(i + 1, n):
            if nums[i] == nums[j]:
                return True
    return False
```

```python
contains_duplicate_brute([1, 2, 3, 1])
```

```python
contains_duplicate_brute([1, 2, 3, 4])
```

```python
contains_duplicate_brute([1, 1, 1, 3, 3, 4, 3, 2, 4, 2])
```

## Solution 2: sort first


Sort first: after sorting, equal values sit next to each other, so one pass over neighbours finds them. The code uses `sorted(nums)`, which builds a sorted copy: O(n) extra memory, but the caller's list is unchanged. If you may change the input, `nums.sort()` sorts in place with little extra memory, which is the version in the table.

```python
def contains_duplicate_sort(nums):            # O(n log n) time
    nums = sorted(nums)                       # use nums.sort() if you may change the input
    for i in range(1, len(nums)):
        if nums[i] == nums[i - 1]:
            return True
    return False
```

```python
sorted([1, 2, 3, 1])
```

```python
contains_duplicate_sort([1, 2, 3, 1])
```

```python
contains_duplicate_sort([1, 2, 3, 4])
```

```python
contains_duplicate_sort([1, 1, 1, 3, 3, 4, 3, 2, 4, 2])
```

## Solution 3: seen-set


Seen-set (the best answer, and the one to present in the interview): one pass, stops at the first duplicate.

Walk through the array once and keep a set of every value seen so far (the seen-set). For each new value, first check the set. If the value is already there, you’ve found a duplicate and can stop. If you reach the end, there were none.

Think of a guest list at a door. Checking whether a name is already on the list is instant, so you never have to re-read the queue.

```python
def contains_duplicate(nums):                 # O(n) time, O(n) space
    seen = set()
    for x in nums:
        if x in seen:
            return True
        seen.add(x)
    return False
```

Always check before adding. If you add first, every value “finds itself” in the set.

Why is a lookup one step? A `dict` or `set` is a hash table. Python runs `hash(x)` to turn the value into a number, and that number picks a slot directly. So `x in seen` jumps to one slot instead of scanning the whole collection. A list has no such shortcut: `x in some_list` checks every element.

Is it always one step? No. It’s one step on average. If many values land in the same slot (a collision), Python has to check each of them, and in the worst case a lookup becomes O(n). This is rare in practice; Python also randomises how strings are hashed so it can’t be triggered on purpose. In an interview, say “O(1) average, O(n) worst case”.

### Walkthrough: `nums = [1, 2, 3, 1]`

| Step | Value | Set before the check | Already in set? | Action |
|---|---|---|---|---|
| 1 | 1 | {} | No | Add 1 |
| 2 | 2 | {1} | No | Add 2 |
| 3 | 3 | {1, 2} | No | Add 3 |
| 4 | 1 | {1, 2, 3} | Yes | Return `true` |

A one-line version also works: `return len(set(nums)) != len(nums)`. It always reads the whole array, even when the duplicate is at the front.

```python
contains_duplicate([1, 2, 3, 1])
```

```python
contains_duplicate([1, 2, 3, 4])
```

```python
contains_duplicate([1, 1, 1, 3, 3, 4, 3, 2, 4, 2])
```

## All three compared


| Approach | Time | Extra memory | When to use it |
|---|---|---|---|
| Brute force | O(n²) | O(1) | Only to state as the starting point |
| Sort first | O(n log n) | Little to none if you sort in place with `nums.sort()` | Memory is tight and you may change the input |
| Seen-set | O(n) | O(n) | Default answer |
