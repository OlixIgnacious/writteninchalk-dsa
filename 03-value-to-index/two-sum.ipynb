## The problem


You are given an integer array `nums` and an integer `target`. Return the indices of the two numbers that add up to `target`.

- Every input has exactly one valid answer.
- You may not use the same element twice.
- The two indices can be returned in any order.

| Input | Output | Why |
|---|---|---|
| `nums = [2, 7, 11, 15]`, `target = 9` | `[0, 1]` | 2 + 7 = 9 |
| `nums = [3, 2, 4]`, `target = 6` | `[1, 2]` | 2 + 4 = 6. Not `[0, 0]`: the single 3 can’t be used twice |
| `nums = [3, 3]`, `target = 6` | `[0, 1]` | Two different elements that happen to be equal |

Input limits: 2 ≤ length of `nums` ≤ 10,000. Values and `target` are between −10⁹ and 10⁹. LeetCode also asks: can you beat O(n²)?

Ask before you code: Do you want indices or the values? Is the array sorted? If it is sorted, you don’t need the dict: two pointers solve it with O(1) memory. That version is a separate problem, Two Sum II (LeetCode 167).

## Solution 1: brute force


Brute force: try every pair.

```python
def two_sum_brute(nums, target):              # O(n^2) time, O(1) space
    for i in range(len(nums)):
        for j in range(i + 1, len(nums)):
            if nums[i] + nums[j] == target:
                return [i, j]
    return []
```

```python
two_sum_brute([3, 2, 4], 6)
```

```python
two_sum_brute([2, 7, 11, 15], 9)
```

```python
two_sum_brute([3, 3], 6)
```

```python
two_sum_brute([-3, 4, 3, 90], 0)
```

```python
two_sum_brute([0, 0], 0)
```

## Solution 2: sort, then two pointers


Sort, then two pointers: sort the positions by their values (not the values themselves, or you lose the original indices). Then move inward from both ends. If the sum is too small, the left value is too small to pair with anything still in range, because even the largest value left, at the right pointer, isn’t enough; so drop it by moving the left pointer right. If the sum is too big, the right value is too big to pair with anything still in range, so drop it by moving the right pointer left. Each move rules out one position that can’t be in the answer, so the pair is never skipped.

```python
def two_sum_sort_two_pointers(nums, target):  # O(n log n) time, O(n) space
    order = sorted(range(len(nums)), key=lambda i: nums[i])   # positions, smallest value first
    l, r = 0, len(nums) - 1
    while l < r:
        s = nums[order[l]] + nums[order[r]]
        if s == target:
            return sorted([order[l], order[r]])
        if s < target:
            l += 1
        else:
            r -= 1
    return []
```

```python
two_sum_sort_two_pointers([3, 2, 4], 6)
```

```python
two_sum_sort_two_pointers([2, 7, 11, 15], 9)
```

```python
two_sum_sort_two_pointers([3, 3], 6)
```

```python
two_sum_sort_two_pointers([-3, 4, 3, 90], 0)
```

```python
two_sum_sort_two_pointers([0, 0], 0)
```

## Solution 3: value-to-index map


Value-to-index map, in one pass (the best answer, and the one to present in the interview):

For each number `x`, the partner you need is `target − x`. Searching the array for the partner every time is slow. Instead, keep a value-to-index map: a dict of the numbers you’ve already passed and the index where you saw each one. Then finding the partner is a single lookup.

Check for the partner before storing `x`. That way `x` can never pair with itself.

Why store the index, not just the value. A set could tell you the partner exists, but problems like Two Sum ask for positions. So the dict maps each value to where you saw it: `{value: index}`.

```python
def two_sum(nums, target):                    # O(n) time, O(n) space
    index_of = {}                             # value -> index
    for i, x in enumerate(nums):
        need = target - x
        if need in index_of:                  # check BEFORE storing x
            return [index_of[need], i]
        index_of[x] = i
    return []
```

You walk the array once (that’s the “one pass”) and store each number with its index. When a number’s partner turns up later, you find it with a single lookup instead of a search.

Why check before storing. If you stored `x` first, it could find itself. With `target = 6` and `x = 3`, the partner is 3, which you’d have just stored. Checking first means the dict only ever holds earlier elements.

```python
two_sum([3, 2, 4], 6)
```

```python
two_sum([2, 7, 11, 15], 9)
```

```python
two_sum([3, 3], 6)
```

```python
two_sum([-3, 4, 3, 90], 0)
```

```python
two_sum([0, 0], 0)
```

## All three compared


| Approach | Time | Extra memory | Note |
|---|---|---|---|
| Brute force | O(n²) | O(1) | Starting point only |
| Sort + two pointers | O(n log n) | O(n) | Needs the position list to keep indices |
| Value-to-index map | O(n) | O(n) | Default answer |
